```python
import numpy as np
import matplotlib.pyplot as plt
np. random. seed (42)

n, p, trials = 10, 0.5, 10000

binomial = np.random.binomial(n, p, trials)

plt.hist(binomial, bins=n + 1, density =True,

alpha=0.7, edgecolor="black")

plt. title(f"Binomial(n=(n), p=(p)): empirical")
plt. show()
```

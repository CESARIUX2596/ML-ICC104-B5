## Homework

Let $X$ and $Y$ be random variables with:

- $\mathbb{E}[X] = 5$, $\mathrm{Var}(X) = 3$
- $\mathbb{E}[Y] = -2$, $\mathrm{Var}(Y) = 7$
- $\mathrm{Cov}(X,Y) = -2$

Compute the following, showing each step:

### a)

$$
\mathbb{E}[2X - 3Y + 1]
$$

### b)

$$
\mathrm{Var}(2X - 3Y + 1)
$$

### c)

$$
\mathrm{Corr}(X,Y)
$$

### d)

If

$$
Z = X + Y
$$

compute:

$$
\mathbb{E}[Z^2]
$$

```python
import math

# Given values
E_X = 5
Var_X = 3
E_Y = -2
Var_Y = 7
Cov_XY = -2

print("--- Homework Results ---")

# a) E[2X - 3Y + 1]
# Linearity of expectation: E[aX + bY + c] = a*E[X] + b*E[Y] + c
res_a = 2 * E_X - 3 * E_Y + 1
print(f"a) E[2X - 3Y + 1] = {res_a}")

# b) Var(2X - 3Y + 1)
# Variance properties: Var(aX + bY + c) = (a^2)*Var(X) + (b^2)*Var(Y) + 2ab*Cov(X,Y)
# Note: The constant term 'c' disappears in variance calculations.
res_b = (2**2) * Var_X + ((-3)**2) * Var_Y + 2 * 2 * (-3) * Cov_XY
print(f"b) Var(2X - 3Y + 1) = {res_b}")

# c) Corr(X, Y)
# Correlation formula: Corr(X, Y) = Cov(X, Y) / (std(X) * std(Y))
std_X = math.sqrt(Var_X)
std_Y = math.sqrt(Var_Y)
res_c = Cov_XY / (std_X * std_Y)
print(f"c) Corr(X, Y) = {res_c:.4f}")

# d) E[Z^2] where Z = X + Y
# Using the formula: E[Z^2] = Var(Z) + (E[Z])^2
E_Z = E_X + E_Y
Var_Z = Var_X + Var_Y + 2 * Cov_XY
res_d = Var_Z + (E_Z**2)
print(f"d) E[Z^2] = {res_d}")
```
